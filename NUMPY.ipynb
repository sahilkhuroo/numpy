## numpy slicing

```python
arr3 = np.array([[12,23,45],[23,34,65],[21,77,37]])
print(arr3[1][2])
```

```python
ar3 = np.array([1,2,3,4,54,6,4,887,22,34,66,88,54,68,34,6])
print(ar3[3:7])
print(ar3[3:])
print(ar3[0:12:2])
print(ar3[::-2])
```

## statical operation

```python
stf = np.array( [1,2,3,4,54,6,4,887,22,34,66,88,54,68,34,65])
print(np.max(stf))  #maximum value
print(np.mean(stf))  #average value
print(np.min(stf))  #minimum value
print(np.sum(stf))  #sum of all item
print(np.median(stf))  #middle value
```

## mathamatical operation

```python
ar1 = np.array([8,7,6])
ar2 = np.array([10,11,12])
np.add(ar1,ar2)
```

```python
np.subtract(ar1,ar2)
```

```python
np.multiply(ar1,ar2)
```

```python
np.divide(ar1,ar2)
```

```python
print(np.power(ar1,2))
```

```python
print(np.sqrt(ar2))
```
